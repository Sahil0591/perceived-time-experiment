# Visual Model — Blind Predictions on Travel Clips

Loads the calibrated visual model from `models/visual_accumulator_latest.json`
and runs predictions on all processed clips in `data/processed/frames/`.

No human data required — uses the pre-calibrated slope, intercept, and compression factor.


```python
# ============================================================================
# SECTION 1: Imports
# ============================================================================

import json
import numpy as np
import pandas as pd
from pathlib import Path
from datetime import datetime

print("Imports loaded")
```

```python
# ============================================================================
# SECTION 2: Load Calibrated Model
# ============================================================================

PROJECT_ROOT = Path.cwd().parent
MODELS_DIR   = PROJECT_ROOT / "models"
FRAMES_DIR   = PROJECT_ROOT / "data" / "processed" / "frames"
OUT_DIR      = PROJECT_ROOT / "results" / "visual" / "travel"
OUT_DIR.mkdir(parents=True, exist_ok=True)

# Load the saved model artifact
model_path = MODELS_DIR / "visual_accumulator_latest.json"
assert model_path.exists(), f"Model not found: {model_path}. Run visual_model.ipynb first."

with open(model_path) as f:
    model = json.load(f)

# Extract calibration and parameters
slope     = model["calibration"]["slope"]
intercept = model["calibration"]["intercept"]
ATTENTION  = model["params"]["ATTENTION"]
EMA_ALPHA  = model["params"]["EMA_ALPHA"]
THRESH_K   = model["params"]["THRESH_K"]
REFRACTORY = model["params"]["REFRACTORY"]
WARMUP     = model["params"]["WARMUP"]

print(f"Model loaded: {model['name']}")
print(f"Created at:   {model['created_at']}")
print(f"Calibration:  slope={slope:.4f}, intercept={intercept:.4f}")
print(f"Params:       ATTENTION={ATTENTION}, THRESH_K={THRESH_K}, WARMUP={WARMUP}")
print(f"Validation:   r={model['metrics']['pearson_r_vs_human']:.3f}, MAE={model['metrics']['mae_vs_human']:.2f}s")
```

```python
# ============================================================================
# SECTION 3: Accumulator Model (identical to visual_model.ipynb)
# ============================================================================

def accumulate_ticks(dist, attention, ema_alpha, thresh_k, refractory, warmup):
    x = (attention * np.asarray(dist, dtype=np.float32)).copy()
    if len(x) > 0 and x[0] == 0:
        x = x[1:]
    if len(x) <= warmup:
        return 0

    mu    = 0.0
    v     = 1e-6
    ticks = 0
    refr  = 0

    for t, xi in enumerate(x):
        xi        = float(xi)
        sigma     = float(np.sqrt(max(v, 1e-8)))
        threshold = mu + thresh_k * sigma

        if t >= warmup:
            if refr > 0:
                refr -= 1
            elif xi > threshold:
                ticks += 1
                refr   = int(refractory)

        prev_mu = mu
        mu = (1.0 - ema_alpha) * mu + ema_alpha * xi
        v  = (1.0 - ema_alpha) * v  + ema_alpha * (xi - prev_mu) * (xi - mu)

    return int(ticks)

print("Accumulator ready")
```

```python
# ============================================================================
# SECTION 4: Run Predictions on All Travel Clips
# ============================================================================

# Find all clip folders — each folder is named by clip ID (1, 2, 3 ... 36)
clip_folders = sorted(
    [d for d in FRAMES_DIR.iterdir() if d.is_dir()],
    key=lambda d: int(d.name) if d.name.isdigit() else 0
)

print(f"Found {len(clip_folders)} clip folder(s) in {FRAMES_DIR}")
print("Running accumulator on each clip...")
print("=" * 60)

results = []

for clip_folder in clip_folders:
    clip_id = clip_folder.name

    # Find the frames CSV for this clip
    csv_files = list(clip_folder.glob("*_frames.csv"))
    if len(csv_files) == 0:
        # Try direct CSV with clip_id name
        csv_files = list(clip_folder.glob("*.csv"))

    if len(csv_files) == 0:
        print(f"  Clip {clip_id}: No CSV found — skipping")
        continue

    # Load L2 distances
    df = pd.read_csv(csv_files[0])

    if "l2_to_prev" not in df.columns:
        print(f"  Clip {clip_id}: Missing l2_to_prev column — skipping")
        continue

    l2_distances = df["l2_to_prev"].values
    duration_s   = float(df["t_sec"].iloc[-1]) if "t_sec" in df.columns and len(df) > 0 else 0.0
    n_frames     = len(df)

    # Run accumulator
    ticks = accumulate_ticks(
        l2_distances,
        ATTENTION, EMA_ALPHA, THRESH_K, REFRACTORY, WARMUP
    )

    # Apply calibration
    pred_s = slope * ticks + intercept

    results.append({
        "clip_id":          clip_id,
        "n_frames":         n_frames,
        "duration_s":       round(duration_s, 2),
        "ticks":            ticks,
        "pred_perceived_s": round(pred_s, 3),
    })

    print(f"  Clip {clip_id:>3}: {n_frames} frames, {duration_s:.1f}s actual → "
          f"{ticks} ticks → {pred_s:.2f}s predicted")

print("=" * 60)
print(f"Done. {len(results)} clips processed.")
```

```python
# ============================================================================
# SECTION 5: Save and Display Results
# ============================================================================

results_df = pd.DataFrame(results)

# Save to CSV
out_path = OUT_DIR / "travel_clip_predictions.csv"
results_df.to_csv(out_path, index=False)
print(f"Predictions saved: {out_path}")

# Save metadata
meta = {
    "model":      model["name"],
    "created_at": model["created_at"],
    "run_at":     datetime.now().isoformat(timespec="seconds"),
    "n_clips":    len(results_df),
    "calibration": model["calibration"],
    "params":      model["params"],
    "summary": {
        "mean_predicted_s": round(float(results_df["pred_perceived_s"].mean()), 2),
        "mean_actual_s":    round(float(results_df["duration_s"].mean()), 2),
        "mean_ticks":       round(float(results_df["ticks"].mean()), 2),
    }
}
with open(OUT_DIR / "travel_predictions_meta.json", "w") as f:
    json.dump(meta, f, indent=2)

# Display results table
print()
print(f"{'Clip':>6} {'Actual (s)':>12} {'Ticks':>8} {'Predicted (s)':>15}")
print("-" * 45)
for _, row in results_df.iterrows():
    print(f"{row['clip_id']:>6} {row['duration_s']:>12.1f} {int(row['ticks']):>8} {row['pred_perceived_s']:>15.2f}")

print("-" * 45)
print(f"{'Mean':>6} {results_df['duration_s'].mean():>12.1f} "
      f"{results_df['ticks'].mean():>8.1f} "
      f"{results_df['pred_perceived_s'].mean():>15.2f}")

print()
print(f"Mean actual duration:    {results_df['duration_s'].mean():.2f}s")
print(f"Mean predicted duration: {results_df['pred_perceived_s'].mean():.2f}s")
```

```python
# ============================================================================
# SECTION 6: Quick Visualisation
# ============================================================================

import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Predicted vs Actual
axes[0].scatter(results_df["duration_s"], results_df["pred_perceived_s"],
                alpha=0.7, color="steelblue", s=60)
max_val = max(results_df["duration_s"].max(), results_df["pred_perceived_s"].max())
axes[0].plot([0, max_val], [0, max_val], "r--", alpha=0.5, label="y = x (perfect)")
axes[0].set_xlabel("Actual Duration (s)")
axes[0].set_ylabel("Predicted Perceived Duration (s)")
axes[0].set_title("Model Predictions vs Actual Duration\n(Travel Clips)")
axes[0].legend()
axes[0].grid(alpha=0.3)

# Tick count distribution
axes[1].bar(results_df["clip_id"].astype(str), results_df["ticks"],
            color="steelblue", alpha=0.7)
axes[1].set_xlabel("Clip ID")
axes[1].set_ylabel("Tick Count")
axes[1].set_title("Accumulator Tick Counts per Clip")
axes[1].tick_params(axis="x", rotation=90)
axes[1].grid(alpha=0.3, axis="y")

plt.tight_layout()
plt.savefig(OUT_DIR / "travel_clip_predictions.png", dpi=150, bbox_inches="tight")
plt.show()
print(f"Plot saved: {OUT_DIR / 'travel_clip_predictions.png'}")
```
