# Visual Model: AlexNet Features and Accumulated Change

This notebook extracts visual features from preprocessed frames using a pretrained AlexNet and computes simple change metrics over time (L2 and cosine distance to the previous frame). It saves per-frame CSVs and a per-clip summary to `results/visual`.

Prerequisites:
- Run `scripts/preprocess_data.py --save-frames` so frames live under `data/processed/frames/<clip_id>/`.
- Install PyTorch and torchvision (CUDA wheels if you want GPU).

```python
# Imports and environment check
import os, math, json
from pathlib import Path
from typing import List, Tuple, Optional, Dict

import numpy as np
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torchvision import transforms as T
from torchvision.models import alexnet, AlexNet_Weights

print(f'Python working dir: {os.getcwd()}')
print('Torch:', torch.__version__)
print('CUDA available:', torch.cuda.is_available())
if torch.cuda.is_available():
    print('CUDA device:', torch.cuda.get_device_name(0))
```

```python
# Configuration (project-root aware)
from pathlib import Path
import os

# VS Code notebooks usually run with cwd = notebooks/, so parent is the repo root
PROJECT_ROOT = Path.cwd().parent
FRAMES_ROOT = PROJECT_ROOT / "data" / "processed" / "frames"
OUT_DIR = PROJECT_ROOT / "results" / "visual"
BATCH_SIZE = 64
FRAME_SIZE = 224  # frames already 224x224 from preprocessing; we enforce and normalize anyway
DEVICE_MODE = 'auto'  # 'auto' | 'cuda' | 'cpu'

FRAMES_ROOT, OUT_DIR
```

```python
# Helpers: device, transforms, model, listing clips
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]

def get_device(mode: str) -> torch.device:
    if mode == 'cuda' and torch.cuda.is_available():
        return torch.device('cuda')
    if mode == 'cpu':
        return torch.device('cpu')
    return torch.device('cuda' if torch.cuda.is_available() else 'cpu')

def build_transform():
    return T.Compose([
        T.Resize((FRAME_SIZE, FRAME_SIZE)),
        T.ToTensor(),
        T.Normalize(mean=IMAGENET_MEAN, std=IMAGENET_STD),
    ])

def build_model(device: torch.device) -> nn.Module:
    weights = AlexNet_Weights.DEFAULT
    base = alexnet(weights=weights).to(device)
    base.eval()
    # Extract up to fc7 (4096-D)
    extractor = nn.Sequential(
        base.features,
        base.avgpool,
        nn.Flatten(),
        nn.Sequential(*list(base.classifier.children())[:6]),
    ).to(device)
    extractor.eval()
    return extractor

def list_clips(frames_root: Path) -> List[Path]:
    if not frames_root.exists():
        return []
    clips = []
    for p in sorted(frames_root.iterdir()):
        if p.is_dir() and (p / 'frame_times.npy').exists():
            clips.append(p)
    return clips

device = get_device(DEVICE_MODE)
print('Using device:', device)
clips = list_clips(FRAMES_ROOT)
print(f'Found {len(clips)} clip(s) under', FRAMES_ROOT)
clips[:5]
```

```python
# Feature extraction and change metrics
def batch_iter(items: List[Path], bs: int):
    for i in range(0, len(items), bs):
        yield items[i:i+bs]

def load_frame_paths(clip_dir: Path) -> List[Path]:
    return sorted([p for p in clip_dir.iterdir() if p.suffix.lower() in {'.jpg', '.jpeg', '.png'}])

def compute_change(emb: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    Tn = emb.shape[0]
    l2 = np.zeros(Tn, dtype=np.float32)
    cos = np.zeros(Tn, dtype=np.float32)
    if Tn <= 1:
        return l2, cos
    a = emb[:-1]
    b = emb[1:]
    diff = b - a
    l2[1:] = np.linalg.norm(diff, axis=1)
    # cosine distance = 1 - cosine similarity
    a_norm = a / (np.linalg.norm(a, axis=1, keepdims=True) + 1e-8)
    b_norm = b / (np.linalg.norm(b, axis=1, keepdims=True) + 1e-8)
    cos_sim = np.sum(a_norm * b_norm, axis=1)
    cos[1:] = 1.0 - np.clip(cos_sim, -1.0, 1.0)
    return l2, cos

def extract_clip(clip_dir: Path, model: nn.Module, tfm, device: torch.device,
                 batch_size: int = 64, save_embeddings: bool = False) -> Dict[str, object]:
    clip_id = clip_dir.name
    img_paths = load_frame_paths(clip_dir)
    if not img_paths:
        raise RuntimeError(f'No frames in {clip_dir}')
    times = np.load(str(clip_dir / 'frame_times.npy')).astype(np.float32)
    assert len(times) == len(img_paths), 'Mismatch frames vs times'

    embs = []
    model.eval()
    with torch.no_grad():
        for batch in batch_iter(img_paths, batch_size):
            imgs = []
            for p in batch:
                with Image.open(p).convert('RGB') as im:
                    imgs.append(tfm(im))
            x = torch.stack(imgs, 0).to(device)
            z = model(x)  # [B, 4096]
            embs.append(z.detach().cpu().numpy())

    emb = np.concatenate(embs, axis=0)
    l2, cos = compute_change(emb)

    duration = float(times[-1] - times[0] + (times[1] - times[0] if len(times) > 1 else 0.0))
    summary = {
        'clip_id': clip_id,
        'n_frames': int(len(times)),
        'duration_sec': duration,
        'sum_l2': float(l2.sum()),
        'mean_l2': float(l2.mean()) if len(l2) else 0.0,
        'sum_cos': float(cos.sum()),
        'mean_cos': float(cos.mean()) if len(cos) else 0.0,
        'rate_l2_per_sec': float(l2.sum() / max(duration, 1e-6)),
    }

    out = {
        'clip_id': clip_id,
        'times': times,
        'l2': l2,
        'cos': cos,
        'embeddings': emb if save_embeddings else None,
        'summary': summary,
    }
    return out
```

```python
# Run extraction over all clips and save outputs
tfm = build_transform()
model = build_model(device)
OUT_DIR.mkdir(parents=True, exist_ok=True)
(OUT_DIR / "frames").mkdir(parents=True, exist_ok=True)

per_clip_summaries = []

for clip_dir in clips:
    clip_id = clip_dir.name
    print("Processing", clip_id)
    res = extract_clip(clip_dir, model, tfm, device, batch_size=BATCH_SIZE, save_embeddings=False)

    df = pd.DataFrame({
        "t_sec": res["times"],
        "frame_idx": np.arange(len(res["times"]), dtype=np.int32),
        "l2_to_prev": res["l2"],
        "cos_to_prev": res["cos"],
    })
    df.to_csv(OUT_DIR / "frames" / f"{clip_id}_frames.csv", index=False)
    per_clip_summaries.append(res["summary"])

summary_df = pd.DataFrame(per_clip_summaries)
summary_df.to_csv(OUT_DIR / "summary.csv", index=False)
print("Wrote summary to", OUT_DIR / "summary.csv")
summary_df.head()
```

```python
# Quick visualization for the first clip (if any)
if len(clips) > 0:
    clip0 = clips[0]
    clip_id = clip0.name
    csv_path = OUT_DIR / "frames" / f"{clip_id}_frames.csv"
    df0 = pd.read_csv(csv_path)

    fig, ax = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
    ax[0].plot(df0["t_sec"], df0["l2_to_prev"], label="L2 to prev")
    ax[0].set_ylabel("L2")
    ax[0].legend()
    ax[1].plot(df0["t_sec"], df0["cos_to_prev"], color="orange", label="Cosine distance")
    ax[1].set_xlabel("Time (s)")
    ax[1].set_ylabel("1 - cos sim")
    ax[1].legend()
    plt.suptitle(f"Visual change over time: {clip_id}")
    plt.show()
else:
    print("No clips found to visualize.")
```

```python
# Baseline: predict perceived time from visual change rate (no human labels)
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

summary_path = OUT_DIR / "summary.csv"
assert summary_path.exists(), f"Missing {summary_path}. Run extraction cells first."
summary = pd.read_csv(summary_path)

# Heuristic: perceived ~ duration * (rate / median_rate)^k
k = 0.75  # tweak 0.5–1.0 for weaker/stronger effect
median_rate = float(summary["rate_l2_per_sec"].median() + 1e-8)
scale = (summary["rate_l2_per_sec"] / median_rate) ** k
proxy = summary["duration_sec"] * scale

# Calibrate so averages match (keeps dataset-level scale realistic)
cal = summary["duration_sec"].mean() / (proxy.mean() + 1e-8)
summary["pred_perceived_sec"] = proxy * cal

out_csv = OUT_DIR / "summary_with_proxy.csv"
summary.to_csv(out_csv, index=False)
print("Wrote:", out_csv)
summary[["clip_id", "duration_sec", "rate_l2_per_sec", "pred_perceived_sec"]].head(10)
```

# Roseboom-style accumulator (closer to the published model)

The proxy above uses a smooth change-rate heuristic. The original Roseboom-style approach instead counts **salient changes**: if the frame-to-frame change exceeds an adaptive threshold, the model increments an **accumulator** (a "tick"). Predicted duration is proportional to the number of ticks.

This section implements a simple single-layer accumulator on your existing `l2_to_prev` series (fc7). It adds an **attention** gain (like your professor’s tuned attention values) and uses an adaptive threshold based on an exponential moving mean/variance.

```python
import numpy as np
import pandas as pd

# Load baseline summary (must exist from earlier extraction cells)
summary_path = OUT_DIR / "summary.csv"
assert summary_path.exists(), f"Missing {summary_path}. Run extraction cells first."
summary = pd.read_csv(summary_path).copy()

# Params (start by matching professor's still tuned attention)
ATTENTION = 0.67
EMA_ALPHA = 0.10
THRESH_K = 1.0
REFRACTORY = 0
WARMUP = 10  # frames used to stabilize threshold (no ticks counted here)

def accumulate_ticks(
    dist: np.ndarray,
    attention: float,
    ema_alpha: float,
    thresh_k: float,
    refractory: int,
    warmup: int = 10,
) -> int:
    """
    Tick when (attention*dist[t]) exceeds an adaptive threshold derived from past distances.
    Threshold uses EMA mean/variance of previous samples.
    """
    x = (attention * np.asarray(dist, dtype=np.float32)).copy()

    # ignore the first element if it's the "no previous frame" 0
    if len(x) > 0 and x[0] == 0:
        x = x[1:]

    mu = 0.0
    v = 1e-6  # small initial variance to avoid zero sigma
    ticks = 0
    refr = 0

    for t, xi in enumerate(x):
        xi = float(xi)

        # threshold from *previous* state
        sigma = float(np.sqrt(max(v, 1e-8)))
        thr = mu + thresh_k * sigma

        # tick decision (skip during warmup)
        if t >= warmup:
            if refr > 0:
                refr -= 1
            else:
                if xi > thr:
                    ticks += 1
                    refr = int(refractory)

        # update EMA mean/var with current sample (after decision)
        prev_mu = mu
        mu = (1.0 - ema_alpha) * mu + ema_alpha * xi
        v = (1.0 - ema_alpha) * v + ema_alpha * (xi - prev_mu) * (xi - mu)

    return int(ticks)

# Compute ticks per clip using per-frame CSVs
ticks_list = []
for clip_id in summary["clip_id"].values:
    frames_csv = OUT_DIR / "frames" / f"{clip_id}_frames.csv"
    df = pd.read_csv(frames_csv)
    ticks_list.append(
        accumulate_ticks(
            df["l2_to_prev"].values,
            attention=ATTENTION,
            ema_alpha=EMA_ALPHA,
            thresh_k=THRESH_K,
            refractory=REFRACTORY,
            warmup=WARMUP,
        )
    )

summary["ticks_fc7"] = ticks_list

# Convert ticks -> seconds by matching dataset mean duration (self-calibration)
mean_duration = float(summary["duration_sec"].mean())
mean_ticks = float(summary["ticks_fc7"].mean() + 1e-8)
sec_per_tick = mean_duration / mean_ticks

summary["sec_per_tick"] = sec_per_tick
summary["pred_perceived_sec_accum"] = summary["ticks_fc7"] * sec_per_tick

out_csv = OUT_DIR / "summary_with_accumulator.csv"
summary.to_csv(out_csv, index=False)
print("Wrote:", out_csv)
summary[["clip_id", "duration_sec", "ticks_fc7", "pred_perceived_sec_accum"]].head(10)
```

# Evaluate on professor's redacted videos (no preprocessing)

Your professor’s reference CSVs use `trial_id` (e.g., 21, 22, …). The redacted MP4s are named like `video_21_redacted.mp4`, so we extract `trial_id=21` from the filename and compare your model’s predictions to the professor’s **still** model estimates.

This section runs AlexNet fc7 extraction directly on the MP4s in `data/redacted/Redacted videos/` and writes outputs to `results/visual/redacted/`.

```python
# Setup: paths, helpers, and extraction function for redacted videos
import re
from pathlib import Path
import numpy as np
import pandas as pd
import cv2
from PIL import Image
from typing import Optional

VIDEO_ROOT = PROJECT_ROOT / "data" / "redacted" / "Redacted videos"
OUT_REDACTED = PROJECT_ROOT / "results" / "visual" / "redacted"
OUT_REDACTED.mkdir(parents=True, exist_ok=True)
(OUT_REDACTED / "frames").mkdir(parents=True, exist_ok=True)

SAMPLE_FPS = 30  # set to None to keep every native frame

# Load trial sequence: Trial -> video_id, start_frame, end_frame
SEQ_PATH = PROJECT_ROOT / "results" / "visual" / "prof" / "Video_trial_sequence(in).csv"
assert SEQ_PATH.exists(), f"Missing trial sequence file: {SEQ_PATH}"
seq = pd.read_csv(SEQ_PATH)
seq.columns = seq.columns.str.strip()
seq = seq.rename(columns={
    "Trial": "trial_id",
    "Video": "video_id",
    "Test_duration_secs": "test_duration_secs",
    "Test_duration_frames": "test_duration_frames",
    "Start_frame": "start_frame",
    "End_frame": "end_frame",
})
print(f"Loaded trial sequence: {len(seq)} rows, {seq['video_id'].nunique()} unique videos")

def list_videos(video_root: Path):
    return sorted([p for p in video_root.iterdir() if p.suffix.lower() in {".mp4", ".mov", ".mkv"}])

def parse_video_id(video_path: Path) -> Optional[int]:
    """Extract the integer video number from filenames like video_9_redacted.mp4 -> 9"""
    m = re.search(r"video_(\d+)_redacted", video_path.stem.lower())
    return int(m.group(1)) if m else None

def extract_from_video(video_path: Path, model, tfm, device, batch_size: int = 64,
                       sample_fps: Optional[int] = 30):
    """Extract AlexNet fc7 features for every sampled frame of the full video."""
    cap = cv2.VideoCapture(str(video_path))
    if not cap.isOpened():
        raise RuntimeError(f"Could not open video: {video_path}")

    fps = float(cap.get(cv2.CAP_PROP_FPS) or 0.0)

    if sample_fps is None or fps <= 0:
        stride = 1
        used_fps = fps if fps > 0 else None
    else:
        stride = max(1, int(round(fps / sample_fps)))
        used_fps = fps / stride if fps > 0 else sample_fps

    times, embs, batch_tensors, batch_times = [], [], [], []
    frame_idx = 0

    model.eval()
    with torch.no_grad():
        while True:
            ok, frame_bgr = cap.read()
            if not ok:
                break
            if frame_idx % stride == 0:
                frame_rgb = cv2.cvtColor(frame_bgr, cv2.COLOR_BGR2RGB)
                im = Image.fromarray(frame_rgb)
                batch_tensors.append(tfm(im))
                batch_times.append(frame_idx / fps if fps > 0 else float(len(times)))
                if len(batch_tensors) >= batch_size:
                    xb = torch.stack(batch_tensors, 0).to(device)
                    embs.append(model(xb).detach().cpu().numpy())
                    times.extend(batch_times)
                    batch_tensors.clear(); batch_times.clear()
            frame_idx += 1
        if batch_tensors:
            xb = torch.stack(batch_tensors, 0).to(device)
            embs.append(model(xb).detach().cpu().numpy())
            times.extend(batch_times)
    cap.release()

    emb = np.concatenate(embs, axis=0) if embs else np.zeros((0, 4096), dtype=np.float32)
    times_arr = np.asarray(times, dtype=np.float32)
    l2, cos = compute_change(emb)
    return {"times": times_arr, "l2": l2, "cos": cos,
            "fps": fps, "used_fps": used_fps, "n_frames_used": len(times_arr)}

videos = list_videos(VIDEO_ROOT)
print(f"Found {len(videos)} redacted video(s)")
# Build a lookup: video_id (int) -> Path
video_map = {parse_video_id(p): p for p in videos if parse_video_id(p) is not None}
print(f"Video IDs available: {sorted(video_map.keys())}")
```

```python
# Extract ALL 33 redacted videos (whole video) and save per-frame CSVs
# Already-extracted videos are skipped automatically.
tfm = build_transform()
model = build_model(device)

video_summaries = []
for video_id, vp in sorted(video_map.items()):
    clip_id = vp.stem
    out_csv = OUT_REDACTED / "frames" / f"{clip_id}_frames.csv"
    if out_csv.exists():
        print(f"Skip {clip_id} (already extracted)")
        # Still read fps for summary
        df_ex = pd.read_csv(out_csv, nrows=2)
        fps_existing = None  # we won't recompute; mark as cached
        video_summaries.append({"video_id": video_id, "clip_id": clip_id, "cached": True})
        continue

    print(f"Processing video_id={video_id} | {clip_id}")
    res = extract_from_video(vp, model, tfm, device, batch_size=BATCH_SIZE, sample_fps=SAMPLE_FPS)

    df = pd.DataFrame({
        "t_sec": res["times"],
        "frame_idx": np.arange(len(res["times"]), dtype=np.int32),
        "l2_to_prev": res["l2"],
        "cos_to_prev": res["cos"],
    })
    df.to_csv(out_csv, index=False)

    video_summaries.append({
        "video_id": video_id,
        "clip_id": clip_id,
        "fps": res["fps"],
        "used_fps": res["used_fps"],
        "n_frames_used": res["n_frames_used"],
        "cached": False,
    })
    print(f"  -> {res['n_frames_used']} frames, fps={res['fps']:.2f}")

print(f"\nDone. {sum(1 for v in video_summaries if not v['cached'])} extracted, "
      f"{sum(1 for v in video_summaries if v['cached'])} cached.")
```

```python
# Cell 12: Accumulator with baseline (intercept) calibration

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
from sklearn.linear_model import LinearRegression

# Load professor's data
prof_path = PROJECT_ROOT / "results" / "visual" / "prof" / "summary_tuned.csv"
assert prof_path.exists(), f"Missing: {prof_path}"
prof = pd.read_csv(prof_path)
prof.columns = prof.columns.str.strip()

print(f"Professor data columns: {prof.columns.tolist()}")
print(f"Professor data shape: {prof.shape}")

# Average across participants per trial
prof_per_trial = prof.groupby("trial_id")[["human_estimates", "still_estimates", "gaze_estimates", "shuffled_estimates", "real_duration"]].mean().reset_index()

# Join trial_id -> video_id, start_frame, end_frame from trial sequence
trials = prof_per_trial.merge(
    seq[["trial_id", "video_id", "start_frame", "end_frame", "test_duration_secs"]],
    on="trial_id", how="inner"
)

# Only keep trials whose video file we actually have
trials = trials[trials["video_id"].isin(video_map.keys())].copy()
print(f"Trials to evaluate: {len(trials)} (covering {trials['video_id'].nunique()} unique videos)")
print(f"Trials with human data: {trials['human_estimates'].notna().sum()}")

# Accumulator parameters
ATTENTION  = 0.67
EMA_ALPHA  = 0.05
THRESH_K   = 1.6
REFRACTORY = 1
WARMUP     = 5

def accumulate_ticks(dist, attention, ema_alpha, thresh_k, refractory, warmup):
    """Count ticks when attention-weighted change exceeds adaptive threshold."""
    x = (attention * np.asarray(dist, dtype=np.float32)).copy()
    if len(x) > 0 and x[0] == 0:
        x = x[1:]
    
    # Handle very short sequences
    if len(x) <= warmup:
        return 0
    
    mu, v, ticks, refr = 0.0, 1e-6, 0, 0
    for t, xi in enumerate(x):
        xi = float(xi)
        thr = mu + thresh_k * float(np.sqrt(max(v, 1e-8)))
        if t >= warmup:
            if refr > 0:
                refr -= 1
            elif xi > thr:
                ticks += 1
                refr = int(refractory)
        prev_mu = mu
        mu = (1.0 - ema_alpha) * mu + ema_alpha * xi
        v  = (1.0 - ema_alpha) * v  + ema_alpha * (xi - prev_mu) * (xi - mu)
    return int(ticks)

# Cache per-video DataFrames to avoid re-reading the same file for many trials
_frame_cache = {}

def get_trial_l2(video_id, start_frame, end_frame):
    """Extract L2 distances for a specific trial segment."""
    clip_id = video_map[video_id].stem
    if video_id not in _frame_cache:
        csv_path = OUT_REDACTED / "frames" / f"{clip_id}_frames.csv"
        if not csv_path.exists():
            print(f"Warning: Missing {csv_path}")
            return np.array([])
        _frame_cache[video_id] = pd.read_csv(csv_path)
    
    df = _frame_cache[video_id]
    
    # Estimate FPS from frame timestamps
    if len(df) > 1:
        fps_est = 1.0 / float(df["t_sec"].iloc[1] - df["t_sec"].iloc[0] + 1e-10)
    else:
        fps_est = 30.0
    
    # Convert frame indices to time
    start_t = start_frame / fps_est
    end_t   = end_frame   / fps_est
    
    # Extract segment with small tolerance
    seg = df[(df["t_sec"] >= start_t - 0.5/fps_est) & (df["t_sec"] <= end_t + 0.5/fps_est)]
    
    return seg["l2_to_prev"].values

# Run accumulator for every trial
print("\nRunning accumulator on trial segments...")
ticks_list = []
for idx, row in trials.iterrows():
    l2 = get_trial_l2(int(row["video_id"]), int(row["start_frame"]), int(row["end_frame"]))
    if len(l2) == 0:
        ticks_list.append(0)
    else:
        ticks_list.append(accumulate_ticks(l2, ATTENTION, EMA_ALPHA, THRESH_K, REFRACTORY, WARMUP))

trials["ticks_fc7"] = ticks_list

# Report zero-tick problem
zero_ticks = (trials["ticks_fc7"] == 0).sum()
print(f"Zero-tick predictions: {zero_ticks} / {len(trials)} ({100*zero_ticks/len(trials):.1f}%)")

# ============================================================================
# CALIBRATION: Linear regression with intercept (baseline time perception)
# ============================================================================

print("\n" + "="*70)
print("CALIBRATION: Linear Regression (ticks → perceived_time)")
print("="*70)

# Filter valid data for calibration
valid_for_calib = ~trials["human_estimates"].isna()
X_calib = trials.loc[valid_for_calib, "ticks_fc7"].values.reshape(-1, 1)
y_calib = trials.loc[valid_for_calib, "human_estimates"].values

print(f"Calibration samples: {len(X_calib)}")
print(f"  Zero-tick trials: {(X_calib.flatten() == 0).sum()}")
print(f"  Non-zero trials: {(X_calib.flatten() > 0).sum()}")

# Fit: perceived_time = slope * ticks + intercept
lr = LinearRegression()
lr.fit(X_calib, y_calib)

slope = lr.coef_[0]
intercept = lr.intercept_

print(f"\nCalibration results:")
print(f"  perceived_time = {slope:.4f} × ticks + {intercept:.4f}")
print(f"  ")
print(f"  Interpretation:")
print(f"    - Slope ({slope:.4f}): seconds per tick")
print(f"    - Intercept ({intercept:.4f}s): baseline perception (even with 0 ticks)")
print(f"  ")
print(f"  Example predictions:")
print(f"    - 0 ticks  → {intercept:.2f}s (baseline)")
print(f"    - 10 ticks → {10*slope + intercept:.2f}s")
print(f"    - 50 ticks → {50*slope + intercept:.2f}s")

# Apply calibration: y = slope * x + intercept
trials["pred_perceived_sec_accum"] = slope * trials["ticks_fc7"] + intercept

# Save intermediate results
out_acc = OUT_REDACTED / "summary_with_accumulator.csv"
trials.to_csv(out_acc, index=False)
print(f"\nWrote: {out_acc}")

# ============================================================================
# EVALUATION: Compare all models against human estimates
# ============================================================================

print("\n" + "="*70)
print("EVALUATION RESULTS - ALL MODELS vs HUMANS")
print("="*70)

# Filter valid predictions
valid_mask = ~np.isnan(trials["human_estimates"])
n_valid = valid_mask.sum()
print(f"\nValid predictions for evaluation: {n_valid} / {len(trials)} ({100*n_valid/len(trials):.1f}%)")

if n_valid == 0:
    print("ERROR: No valid predictions to evaluate!")
else:
    # Extract valid data
    x_valid = trials.loc[valid_mask, "pred_perceived_sec_accum"].values
    y_valid = trials.loc[valid_mask, "human_estimates"].values
    still_valid = trials.loc[valid_mask, "still_estimates"].values
    gaze_valid = trials.loc[valid_mask, "gaze_estimates"].values
    shuffled_valid = trials.loc[valid_mask, "shuffled_estimates"].values
    real_valid = trials.loc[valid_mask, "real_duration"].values
    
    # Evaluate all models
    models = {
        "Your Model": x_valid,
        "Still (Full-frame)": still_valid,
        "Gaze (Attention)": gaze_valid,
        "Shuffled (Control)": shuffled_valid
    }
    
    results = {}
    for model_name, predictions in models.items():
        r = pearsonr(predictions, y_valid)[0]
        mae = np.abs(predictions - y_valid).mean()
        rmse = np.sqrt(((predictions - y_valid)**2).mean())
        r_real = pearsonr(predictions, real_valid)[0]
        results[model_name] = {
            'r': r,
            'mae': mae,
            'rmse': rmse,
            'r_real': r_real
        }
    
    # Display results in a table
    print(f"\n{'Model':<25} {'Pearson r':<12} {'MAE':<10} {'RMSE':<10} {'r vs Real':<10}")
    print("="*75)
    for model_name, metrics in sorted(results.items(), key=lambda x: x[1]['r'], reverse=True):
        print(f"{model_name:<25} {metrics['r']:>8.3f}    {metrics['mae']:>7.2f}s  {metrics['rmse']:>7.2f}s  {metrics['r_real']:>7.3f}")
    
    # Humans vs real duration (baseline reference)
    r_human_real = pearsonr(y_valid, real_valid)[0]
    print(f"\n{'Human Participants':<25} {'--':<12} {'--':<10} {'--':<10} {r_human_real:>7.3f}")
    
    # Improvement analysis
    print(f"\n" + "="*75)
    print("IMPROVEMENT FROM BASELINE (intercept) CALIBRATION")
    print("="*75)
    
    # Check zero-tick performance specifically
    zero_mask = (trials["ticks_fc7"] == 0) & valid_mask
    if zero_mask.sum() > 0:
        print(f"\nZero-tick trials (n={zero_mask.sum()}):")
        print(f"  Mean human estimate: {trials.loc[zero_mask, 'human_estimates'].mean():.2f}s")
        print(f"  Your model (with intercept): {trials.loc[zero_mask, 'pred_perceived_sec_accum'].mean():.2f}s")
        print(f"  MAE: {np.abs(trials.loc[zero_mask, 'pred_perceived_sec_accum'] - trials.loc[zero_mask, 'human_estimates']).mean():.2f}s")
        print(f"  ✅ Instead of predicting 0s, model now predicts ~{intercept:.2f}s baseline")
    
    # Sanity checks
    print(f"\n" + "="*75)
    print("SANITY CHECKS")
    print("="*75)
    
    your_r_real = results["Your Model"]["r_real"]
    if your_r_real > 0.90:
        print(f"\n⚠️  WARNING: Your model (r={your_r_real:.3f} vs real) is too correlated with physical time!")
        print(f"    Consider increasing THRESH_K or ATTENTION to make it less sensitive.")
    elif your_r_real < 0.40:
        print(f"\n⚠️  WARNING: Your model (r={your_r_real:.3f} vs real) shows weak correlation with physical time!")
        print(f"    Consider decreasing THRESH_K or increasing WARMUP.")
    else:
        print(f"\n✅ Your model (r={your_r_real:.3f} vs real) shows appropriate correlation with physical time.")
    
    # Performance ranking
    print(f"\n" + "="*75)
    print("PERFORMANCE RANKING (by correlation with humans)")
    print("="*75)
    
    ranked = sorted(results.items(), key=lambda x: x[1]['r'], reverse=True)
    for rank, (model_name, metrics) in enumerate(ranked, 1):
        emoji = "🥇" if rank == 1 else "🥈" if rank == 2 else "🥉" if rank == 3 else "  "
        print(f"{emoji} {rank}. {model_name:<25} r={metrics['r']:.3f}  MAE={metrics['mae']:.2f}s")
    
    # Bias analysis by duration ranges
    print(f"\n" + "="*75)
    print("BIAS ANALYSIS: Time Compression/Expansion Patterns")
    print("="*75)
    
    duration_ranges = [
        ("Short (< 10s)", trials["real_duration"] < 10),
        ("Medium (10-40s)", (trials["real_duration"] >= 10) & (trials["real_duration"] <= 40)),
        ("Long (> 40s)", trials["real_duration"] > 40)
    ]
    
    for range_name, range_mask in duration_ranges:
        combined_mask = range_mask & valid_mask
        if combined_mask.sum() > 0:
            subset = trials[combined_mask]
            n = combined_mask.sum()
            real_mean = subset['real_duration'].mean()
            human_mean = subset['human_estimates'].mean()
            your_mean = subset['pred_perceived_sec_accum'].mean()
            still_mean = subset['still_estimates'].mean()
            gaze_mean = subset['gaze_estimates'].mean()
            shuffled_mean = subset['shuffled_estimates'].mean()
            
            print(f"\n{range_name}: n={n}")
            print(f"  Real duration (mean):      {real_mean:>6.2f}s")
            print(f"  Human estimate:            {human_mean:>6.2f}s  [{human_mean/real_mean:>4.2f}x real]")
            print(f"  Your Model:                {your_mean:>6.2f}s  [{your_mean/real_mean:>4.2f}x real]")
            print(f"  Still (Full-frame):        {still_mean:>6.2f}s  [{still_mean/real_mean:>4.2f}x real]")
            print(f"  Gaze (Attention):          {gaze_mean:>6.2f}s  [{gaze_mean/real_mean:>4.2f}x real]")
            print(f"  Shuffled (Control):        {shuffled_mean:>6.2f}s  [{shuffled_mean/real_mean:>4.2f}x real]")
    
    # Visualize - 2x3 grid comparing all models
    fig, axes = plt.subplots(2, 3, figsize=(18, 12))
    axes = axes.flatten()
    
    plot_configs = [
        ("Your Model", "Your Model", x_valid, 'blue'),
        ("Still (Full-frame)", "Still (Full-frame)", still_valid, 'green'),
        ("Gaze (Attention)", "Gaze (Attention)", gaze_valid, 'orange'),
        ("Shuffled (Control)", "Shuffled (Control)", shuffled_valid, 'red'),
        ("Your vs Still", None, (x_valid, still_valid), 'purple'),
        ("Your vs Gaze", None, (x_valid, gaze_valid), 'brown')
    ]
    
    for idx, config in enumerate(plot_configs):
        ax = axes[idx]
        
        if idx < 4:  # Model vs Humans
            title, model_key, predictions, color = config
            ax.scatter(predictions, y_valid, alpha=0.4, s=15, color=color)
            max_val = max(predictions.max(), y_valid.max())
            ax.plot([0, max_val], [0, max_val], 'r--', linewidth=2, label='y=x')
            ax.set_xlabel(f'{title} Prediction (s)', fontsize=10)
            ax.set_ylabel('Human Estimates (s)', fontsize=10)
            r = results[model_key]['r']
            mae = results[model_key]['mae']
            ax.set_title(f'{title} vs Humans\nr={r:.3f}, MAE={mae:.1f}s', fontsize=11)
        else:  # Model vs Model
            title, _, data, color = config
            x_data, y_data = data
            ax.scatter(x_data, y_data, alpha=0.4, s=15, color=color)
            max_val = max(x_data.max(), y_data.max())
            ax.plot([0, max_val], [0, max_val], 'r--', linewidth=2, label='y=x')
            r_compare = pearsonr(x_data, y_data)[0]
            
            if "Still" in title:
                ax.set_xlabel('Your Model Prediction (s)', fontsize=10)
                ax.set_ylabel('Still Model Prediction (s)', fontsize=10)
            else:
                ax.set_xlabel('Your Model Prediction (s)', fontsize=10)
                ax.set_ylabel('Gaze Model Prediction (s)', fontsize=10)
            
            ax.set_title(f'{title}\nr={r_compare:.3f}', fontsize=11)
        
        ax.legend(fontsize=9)
        ax.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plot_path = OUT_REDACTED / "comparison_all_models.png"
    plt.savefig(plot_path, dpi=150, bbox_inches='tight')
    print(f"\nSaved comprehensive plot: {plot_path}")
    plt.show()

# Save final comparison with all model predictions
comparison = trials[["trial_id", "human_estimates", "still_estimates", "gaze_estimates", 
                     "shuffled_estimates", "real_duration", "video_id", "start_frame", 
                     "end_frame", "test_duration_secs", "ticks_fc7", "pred_perceived_sec_accum"]].copy()
out_comparison = OUT_REDACTED / "comparison_all_models.csv"
comparison.to_csv(out_comparison, index=False)
print(f"\nWrote: {out_comparison}")

print("\n" + "="*75)
print("✅ DONE - Model now includes baseline perception (intercept)")
print("="*75)
```
